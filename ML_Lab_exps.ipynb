```python
# -----------------------------
# FIND-S Algorithm
# -----------------------------

import pandas as pd

# Training Dataset
data = [
    ['Sunny','Warm','Normal','Strong','Warm','Same','Yes'],
    ['Sunny','Warm','High','Strong','Warm','Same','Yes'],
    ['Rainy','Cold','High','Strong','Warm','Change','No'],
    ['Sunny','Warm','High','Strong','Cool','Change','Yes']
]

columns = ['Sky','AirTemp','Humidity','Wind','Water','Forecast','EnjoySport']

df = pd.DataFrame(data, columns=columns)

print("Training Dataset\n")
print(df)

# -----------------------------
# FIND-S Algorithm
# -----------------------------

def find_s_algorithm(data):

    # Select only positive examples
    positive_examples = data[data['EnjoySport'] == 'Yes']

    # Initialize hypothesis
    hypothesis = positive_examples.iloc[0, :-1].tolist()

    print("\nInitial Hypothesis:")
    print(hypothesis)

    # Iterate through positive examples
    for i in range(1, len(positive_examples)):

        example = positive_examples.iloc[i, :-1].tolist()

        print("\nComparing with:", example)

        for j in range(len(hypothesis)):

            if hypothesis[j] != example[j]:
                hypothesis[j] = '?'

        print("Updated Hypothesis:", hypothesis)

    return hypothesis

# Run FIND-S
final_hypothesis = find_s_algorithm(df)

print("\n--------------------------------")
print("Final Hypothesis")
print("--------------------------------")
print(final_hypothesis)
```

```python
# ============================================
# Candidate Elimination Algorithm
# ============================================

import pandas as pd
import numpy as np

# --------------------------------------------
# Create CSV File
# --------------------------------------------

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("trainingdata.csv", index=False)

print("CSV File Created Successfully\n")

# --------------------------------------------
# Read CSV File
# --------------------------------------------

dataset = pd.read_csv("trainingdata.csv")

print("Training Dataset\n")
print(dataset)

# --------------------------------------------
# Separate Features and Target
# --------------------------------------------

concepts = np.array(dataset.iloc[:, :-1])
target = np.array(dataset.iloc[:, -1])

# --------------------------------------------
# Candidate Elimination Algorithm
# --------------------------------------------

def candidate_elimination(concepts, target):

    specific = concepts[0].copy()

    general = [["?" for i in range(len(specific))]
               for j in range(len(specific))]

    print("\nInitial Specific Hypothesis:")
    print(specific)

    print("\nInitial General Hypothesis:")
    for g in general:
        print(g)

    for i in range(len(concepts)):

        if target[i] == "Yes":

            for j in range(len(specific)):

                if concepts[i][j] != specific[j]:

                    specific[j] = "?"
                    general[j][j] = "?"

        else:

            for j in range(len(specific)):

                if concepts[i][j] != specific[j]:

                    general[j][j] = specific[j]

                else:

                    general[j][j] = "?"

        print("\n--------------------------------")
        print("Step", i + 1)

        print("Specific Hypothesis")
        print(specific)

        print("\nGeneral Hypothesis")

        for g in general:
            print(g)

    # Remove rows having all '?'
    final_general = []

    for g in general:
        if g != ["?"] * len(specific):
            final_general.append(g)

    return specific, final_general


# --------------------------------------------
# Run Algorithm
# --------------------------------------------

specific, general = candidate_elimination(concepts, target)

print("\n===================================")
print("Final Specific Hypothesis")
print("===================================")
print(specific)

print("\n===================================")
print("Final General Hypothesis")
print("===================================")

for g in general:
    print(g)
```

```python
# ===========================================
# Decision Tree (ID3 Algorithm)
# Google Colab Program
# ===========================================

import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

print("=====================================")
print("Decision Tree (ID3 Algorithm)")
print("=====================================")

# -----------------------------
# User Input
# -----------------------------

n = int(input("Enter Number of Training Samples: "))

data = []

print("\nEnter the training data")

for i in range(n):

    print(f"\nSample {i+1}")

    outlook = input("Outlook (Sunny/Overcast/Rain): ")
    temperature = input("Temperature (Hot/Mild/Cool): ")
    humidity = input("Humidity (High/Normal): ")
    wind = input("Wind (Weak/Strong): ")
    play = input("Play Tennis (Yes/No): ")

    data.append([outlook, temperature, humidity, wind, play])

# ------------------------------------
# Create DataFrame
# ------------------------------------

columns = ["Outlook","Temperature","Humidity","Wind","Play"]

df = pd.DataFrame(data, columns=columns)

print("\nTraining Dataset\n")
print(df)

# ------------------------------------
# Convert Categorical Data into Numbers
# ------------------------------------

df_encoded = df.copy()

encoding = {}

for column in df_encoded.columns:

    unique_values = df_encoded[column].unique()

    mapping = {value:index for index,value in enumerate(unique_values)}

    encoding[column] = mapping

    df_encoded[column] = df_encoded[column].map(mapping)

# ------------------------------------
# Features and Target
# ------------------------------------

X = df_encoded.iloc[:,:-1]

y = df_encoded.iloc[:,-1]

# ------------------------------------
# Train Decision Tree (ID3)
# ------------------------------------

model = DecisionTreeClassifier(criterion="entropy")

model.fit(X,y)

# ------------------------------------
# Display Decision Tree
# ------------------------------------

print("\n=====================================")
print("Decision Tree")
print("=====================================")

tree_rules = export_text(model, feature_names=list(X.columns))

print(tree_rules)

# ------------------------------------
# Predict New Sample
# ------------------------------------

print("\nEnter New Sample")

outlook_options = "/".join(list(encoding["Outlook"].keys()))
temperature_options = "/".join(list(encoding["Temperature"].keys()))
humidity_options = "/".join(list(encoding["Humidity"].keys()))
wind_options = "/".join(list(encoding["Wind"].keys()))


# Loop until valid inputs are provided or user quits
while True:
    try:
        outlook = input(f"Outlook ({outlook_options}): ")
        temperature = input(f"Temperature ({temperature_options}): ")
        humidity = input(f"Humidity ({humidity_options}): ")
        wind = input(f"Wind ({wind_options}): ")

        sample = [
            encoding["Outlook"][outlook],
            encoding["Temperature"][temperature],
            encoding["Humidity"][humidity],
            encoding["Wind"][wind]
        ]
        break # Exit loop if all inputs are valid
    except KeyError as e:
        print(f"Error: Invalid input for {e}. Please choose from the available options.")
        print("Please re-enter the sample details.")

prediction = model.predict([sample])

reverse = {v:k for k,v in encoding["Play"].items()}

print("\n=====================================")
print("Prediction")
print("=====================================")

print("Class =", reverse[prediction[0]])
```

```python
# ============================================
# Artificial Neural Network using Backpropagation
# Google Colab Code
# ============================================

import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import LabelEncoder

print("="*50)
print("Artificial Neural Network using Backpropagation")
print("="*50)

# -----------------------------
# Number of Training Samples
# -----------------------------
n = int(input("Enter Number of Training Samples: "))

X = []
y = []

print("\nEnter Training Data")

for i in range(n):
    print(f"\nSample {i+1}")

    f1 = float(input("Feature 1: "))
    f2 = float(input("Feature 2: "))
    f3 = float(input("Feature 3: "))
    target = input("Class (Yes/No): ")

    X.append([f1, f2, f3])
    y.append(target)

# -----------------------------
# Encode Output Labels
# -----------------------------

encoder = LabelEncoder()
y = encoder.fit_transform(y)

X = np.array(X)

# -----------------------------
# Build ANN Model
# -----------------------------

model = MLPClassifier(
    hidden_layer_sizes=(5,),
    activation='logistic',
    solver='sgd',
    learning_rate_init=0.1,
    max_iter=5000,
    random_state=1
)

# -----------------------------
# Train Model
# -----------------------------

model.fit(X, y)

print("\nModel Training Completed Successfully!")

# -----------------------------
# Training Accuracy
# -----------------------------

accuracy = model.score(X, y)

print("\nTraining Accuracy = {:.2f}%".format(accuracy*100))

# -----------------------------
# Test New Sample
# -----------------------------

print("\nEnter Test Sample")

t1 = float(input("Feature 1: "))
t2 = float(input("Feature 2: "))
t3 = float(input("Feature 3: "))

prediction = model.predict([[t1,t2,t3]])

print("\nPrediction =", encoder.inverse_transform(prediction)[0])
```

```python
# ============================================
# K-Nearest Neighbours (K-NN) Algorithm
# Google Colab Code
# ============================================

import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder

print("="*50)
print("K-Nearest Neighbours (K-NN)")
print("="*50)

# -----------------------------
# Input Training Data
# -----------------------------

n = int(input("Enter Number of Training Samples: "))

X = []
y = []

print("\nEnter Training Data")

for i in range(n):

    print(f"\nSample {i+1}")

    f1 = float(input("Feature 1: "))
    f2 = float(input("Feature 2: "))
    target = input("Class (Yes/No): ")

    X.append([f1, f2])
    y.append(target)

# -----------------------------
# Encode Class Labels
# -----------------------------

encoder = LabelEncoder()
y = encoder.fit_transform(y)

X = np.array(X)

# -----------------------------
# Value of K
# -----------------------------

k = int(input("\nEnter Value of K: "))

# -----------------------------
# Train KNN Model
# -----------------------------

model = KNeighborsClassifier(n_neighbors=k)

model.fit(X, y)

print("\nModel Trained Successfully!")

# -----------------------------
# Training Accuracy
# -----------------------------

accuracy = model.score(X, y)

print("Training Accuracy = {:.2f}%".format(accuracy * 100))

# -----------------------------
# Predict New Sample
# -----------------------------

print("\nEnter Test Sample")

t1 = float(input("Feature 1: "))
t2 = float(input("Feature 2: "))

prediction = model.predict([[t1, t2]])

print("\nPrediction =", encoder.inverse_transform(prediction)[0])

# -----------------------------
# Nearest Neighbours
# -----------------------------

distances, indices = model.kneighbors([[t1, t2]])

print("\nNearest Neighbour Indices:")
print(indices)

print("\nDistances:")
print(distances)
```

```python
# ============================================
# Naive Bayes Algorithm
# Google Colab Code
# ============================================

import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

print("="*50)
print("Naive Bayes Algorithm")
print("="*50)

# -----------------------------
# Training Data Input
# -----------------------------

n = int(input("Enter Number of Training Samples: "))

X_train = []
y_train = []

print("\nEnter Training Data")

for i in range(n):

    print(f"\nSample {i+1}")

    f1 = float(input("Feature 1: "))
    f2 = float(input("Feature 2: "))
    target = input("Class (Yes/No): ")

    X_train.append([f1, f2])
    y_train.append(target)

# -----------------------------
# Encode Target Labels
# -----------------------------

encoder = LabelEncoder()

y_train = encoder.fit_transform(y_train)

X_train = np.array(X_train)

# -----------------------------
# Train Naive Bayes Model
# -----------------------------

model = GaussianNB()

model.fit(X_train, y_train)

print("\nModel Trained Successfully!")

# -----------------------------
# Test Data Input
# -----------------------------

m = int(input("\nEnter Number of Test Samples: "))

X_test = []
y_test = []

print("\nEnter Test Data")

for i in range(m):

    print(f"\nTest Sample {i+1}")

    f1 = float(input("Feature 1: "))
    f2 = float(input("Feature 2: "))
    target = input("Actual Class (Yes/No): ")

    X_test.append([f1, f2])
    y_test.append(target)

X_test = np.array(X_test)
y_test = encoder.transform(y_test)

# -----------------------------
# Prediction
# -----------------------------

y_pred = model.predict(X_test)

# -----------------------------
# Display Predictions
# -----------------------------

print("\nPredicted Classes:")

for i in range(m):
    print(f"Sample {i+1} --> {encoder.inverse_transform([y_pred[i]])[0]}")

# -----------------------------
# Confusion Matrix
# -----------------------------

cm = confusion_matrix(y_test, y_pred)

print("\n================================")
print("Confusion Matrix")
print("================================")
print(cm)

# -----------------------------
# Accuracy
# -----------------------------

acc = accuracy_score(y_test, y_pred)

print("\nAccuracy = {:.2f}%".format(acc*100))

# -----------------------------
# Classification Report
# -----------------------------

print("\nClassification Report")
print(classification_report(y_test, y_pred,
      target_names=encoder.classes_))
```

```python
# ============================================
# Linear Regression Algorithm
# Google Colab Code
# ============================================

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

print("="*50)
print("Linear Regression Algorithm")
print("="*50)

# -----------------------------
# Training Data
# -----------------------------

n = int(input("Enter Number of Training Samples: "))

X_train = []
y_train = []

print("\nEnter Training Data")

for i in range(n):

    print(f"\nSample {i+1}")

    x = float(input("Input Feature (X): "))
    y = float(input("Target Value (Y): "))

    X_train.append([x])
    y_train.append(y)

X_train = np.array(X_train)
y_train = np.array(y_train)

# -----------------------------
# Train Linear Regression Model
# -----------------------------

model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel Trained Successfully!")

# -----------------------------
# Model Parameters
# -----------------------------

print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

# -----------------------------
# Testing Data
# -----------------------------

m = int(input("\nEnter Number of Test Samples: "))

X_test = []
y_test = []

print("\nEnter Test Data")

for i in range(m):

    print(f"\nTest Sample {i+1}")

    x = float(input("Input Feature (X): "))
    y = float(input("Actual Target Value (Y): "))

    X_test.append([x])
    y_test.append(y)

X_test = np.array(X_test)
y_test = np.array(y_test)

# -----------------------------
# Prediction
# -----------------------------

y_pred = model.predict(X_test)

# -----------------------------
# Display Predictions
# -----------------------------

print("\n==============================")
print("Predicted Values")
print("==============================")

for i in range(m):
    print("Input =", X_test[i][0],
          " Predicted =", round(y_pred[i],2),
          " Actual =", y_test[i])

# -----------------------------
# Evaluation
# -----------------------------

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\n==============================")
print("Model Evaluation")
print("==============================")

print("Mean Squared Error (MSE):", round(mse,4))
print("R2 Score:", round(r2,4))
```

```python
# ============================================
# Compare Linear and Polynomial Regression
# Google Colab Program
# ============================================

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

print("="*60)
print("Compare Linear and Polynomial Regression")
print("="*60)

# -----------------------------
# Training Data Input
# -----------------------------

n = int(input("Enter Number of Training Samples: "))

X = []
y = []

print("\nEnter Training Data")

for i in range(n):
    print(f"\nSample {i+1}")
    x = float(input("Input Feature (X): "))
    target = float(input("Target Value (Y): "))

    X.append([x])
    y.append(target)

X = np.array(X)
y = np.array(y)

# -----------------------------
# Linear Regression
# -----------------------------

linear_model = LinearRegression()
linear_model.fit(X, y)

linear_pred = linear_model.predict(X)

linear_mse = mean_squared_error(y, linear_pred)
linear_r2 = r2_score(y, linear_pred)

# -----------------------------
# Polynomial Regression
# -----------------------------

degree = int(input("\nEnter Polynomial Degree (2 or 3): "))

poly = PolynomialFeatures(degree=degree)

X_poly = poly.fit_transform(X)

poly_model = LinearRegression()

poly_model.fit(X_poly, y)

poly_pred = poly_model.predict(X_poly)

poly_mse = mean_squared_error(y, poly_pred)
poly_r2 = r2_score(y, poly_pred)

# -----------------------------
# Results
# -----------------------------

print("\n==============================")
print("Linear Regression")
print("==============================")
print("Slope:", linear_model.coef_[0])
print("Intercept:", linear_model.intercept_)
print("MSE:", round(linear_mse,4))
print("R2 Score:", round(linear_r2,4))

print("\n==============================")
print("Polynomial Regression")
print("==============================")
print("Degree:", degree)
print("Coefficients:", poly_model.coef_)
print("Intercept:", poly_model.intercept_)
print("MSE:", round(poly_mse,4))
print("R2 Score:", round(poly_r2,4))

# -----------------------------
# Compare Models
# -----------------------------

print("\n==============================")
print("Comparison")
print("==============================")

if poly_r2 > linear_r2:
    print("Polynomial Regression performs better.")
elif poly_r2 < linear_r2:
    print("Linear Regression performs better.")
else:
    print("Both models perform equally.")
```

```python
# ============================================
# Expectation Maximization (EM) Algorithm
# Google Colab Program
# ============================================

import numpy as np
from sklearn.mixture import GaussianMixture

print("="*55)
print("Expectation Maximization (EM) Algorithm")
print("="*55)

# -----------------------------
# Input Training Data
# -----------------------------

n = int(input("Enter Number of Data Points: "))

X = []

print("\nEnter Data Points")

for i in range(n):
    print(f"\nPoint {i+1}")
    x = float(input("Feature 1: "))
    y = float(input("Feature 2: "))
    X.append([x, y])

X = np.array(X)

# -----------------------------
# Number of Clusters
# -----------------------------

k = int(input("\nEnter Number of Clusters: "))

# -----------------------------
# Train Gaussian Mixture Model
# -----------------------------

gmm = GaussianMixture(
    n_components=k,
    random_state=42
)

gmm.fit(X)

# -----------------------------
# Predict Clusters
# -----------------------------

labels = gmm.predict(X)

# -----------------------------
# Results
# -----------------------------

print("\n==============================")
print("Cluster Assignment")
print("==============================")

for i in range(n):
    print(f"Point {X[i]} --> Cluster {labels[i]}")

print("\n==============================")
print("Cluster Means")
print("==============================")
print(gmm.means_)

print("\n==============================")
print("Cluster Weights")
print("==============================")
print(gmm.weights_)

print("\n==============================")
print("Covariance Matrices")
print("==============================")
print(gmm.covariances_)
```
